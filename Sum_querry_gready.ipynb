```python
def  kin_energy(x):
    f=np.unique(x,return_counts=True)
    prob=f[1]/x.shape[0]
    energy=np.sum(prob**2)
    return  energy
```

```python
def count_common(x):
    a=np.array(x)
    b=np.where(a==np.min(a))
    v=np.array(b)
    nr_ap=v.shape[1]
    return nr_ap
```

```python
def count_least(x):
    a=np.array(x)
    b=np.where(a==np.max(a))
    v=np.array(b)
    nr_ap=v.shape[1]
    return nr_ap
```

```python
def gcd(a, b):
    if a == 0 or b == 0:
        return max(a, b)
    return gcd(b, a % b)
```

```python
X["GCD_FEAT"]=X.apply(lambda r: gcd(r["feature_elusive_vapoury_accomplice"],r["feature_fortified_repressed_baird"]),axis=1)
```

```python
def l(a, b):
    if a==0 or b==0:
        return max(a, b)
    else:
        if a!=0 and b!=0:
           lcm=(a*b)/gcd(a, b)
    if a==0 and b==0:
        return 0
    return lcm
```

```python
X["LCM_FEAT"]=X.apply(lambda r: l(r["feature_elusive_vapoury_accomplice"], r["feature_limiest_heliolithic_york"]),axis=1)
```

```python
def sum_q(l, r, x):
   
    a=np.array(x)
    n=len(a)
    if l>r:
        aux=l
        l=r
        r=aux
        s=sum(a[i] for i in range(l, r))
        return s   
    if r==l:
        return a[r]
    if n>=r and  n>l:
            s=sum(a[i] for i in range(l, r))
  
            return s   
```

```python
def money(x):
    a=0
    while x>0:
        if x>=10:
            x=x-10
        else: 
            if x>=5:
                x=x-5
            else:
                x=x-1
            a=a+1
    return a
```
